```python
import fitz  # PyMuPDF
from pathlib import Path
```

```python
pdf_path = Path("../data/ct200_manual.pdf")

doc = fitz.open(pdf_path)

print("Pages:", len(doc))
print("Metadata:", doc.metadata)
```

```python
page = doc[0]

print(page)
```

```python
text = page.get_text()

print(text)
```

```python
for page_num, page in enumerate(doc):

    blocks = page.get_text("dict")["blocks"]

    print(f"Page {page_num + 1}: {len(blocks)} blocks")
```

```python
from pprint import pprint

pprint(blocks[0])
```

```python
font_sizes = {}

for page in doc:

    blocks = page.get_text("dict")["blocks"]

    for block in blocks:

        if "lines" not in block:
            continue

        for line in block["lines"]:

            for span in line["spans"]:

                size = span["size"]

                if size not in font_sizes:
                    font_sizes[size] = 0

                font_sizes[size] += 1

print(font_sizes)
```

```python
for page in doc:

    blocks = page.get_text("dict")["blocks"]

    for block in blocks:

        if "lines" not in block:
            continue

        for line in block["lines"]:

            for span in line["spans"]:

                print(
                    f"Size: {span['size']:<6} | "
                    f"Font: {span['font']:<20} | "
                    f"Text: {span['text']}"
                )
```
